# BASE COMISSÃO FINAL

```python
import pandas as pd
import numpy as np
from datetime import datetime

# Importar BASE_COMMIS_FINAL
base_comissao = pd.read_excel('../BASE_COMMISS_FINAL.xlsx')
print("Dados carregados com sucesso!")
print(f"Dimensões: {base_comissao.shape}")
print(base_comissao.head())
```

```python
# Filtrar por marca - Retornar todas as comissões de uma marca específica (ex: PRETO, BRANCO)
marca_filtro = 'PRETO'  # Altere conforme necessário
comissoes_marca = base_comissao[base_comissao['Descr_Marca'] == marca_filtro]
print(f"Comissões da marca {marca_filtro}:")
print(comissoes_marca[['Descr_Marca', 'Descri_Cargo', '%_Comiss']])
print(f"\nTotal de cargos com essa marca: {len(comissoes_marca)}")
```

```python
# Filtrar por cargo - Listar todos os cargos disponíveis e suas comissões (ex: todos VENDEDOR LOJA)
cargo_filtro = 'VENDEDOR LOJA'  # Altere conforme necessário
comissoes_cargo = base_comissao[base_comissao['Descri_Cargo'] == cargo_filtro]
print(f"Comissões para o cargo {cargo_filtro}:")
print(comissoes_cargo[['Descr_Marca', 'Descri_Cargo', '%_Comiss']].sort_values('%_Comiss', ascending=False))
print(f"\nTotal de marcas com esse cargo: {len(comissoes_cargo)}")
```

```python
# Encontrar comissão máxima - Qual é a maior percentagem de comissão na tabela?
# Encontrar comissão mínima - Qual cargo/marca tem a menor comissão?
comissao_maxima = base_comissao.loc[base_comissao['%_Comiss'].idxmax()]
comissao_minima = base_comissao.loc[base_comissao['%_Comiss'].idxmin()]

print("COMISSÃO MÁXIMA:")
print(f"Marca: {comissao_maxima['Descr_Marca']}")
print(f"Cargo: {comissao_maxima['Descri_Cargo']}")
print(f"Percentual: {comissao_maxima['%_Comiss']}%\n")

print("COMISSÃO MÍNIMA:")
print(f"Marca: {comissao_minima['Descr_Marca']}")
print(f"Cargo: {comissao_minima['Descri_Cargo']}")
print(f"Percentual: {comissao_minima['%_Comiss']}%")
```

```python
# Contar registros - Quantas linhas (combinações marca-cargo) existem no total?
total_registros = len(base_comissao)
print(f"Total de combinações marca-cargo: {total_registros}")
print(f"\nDistribuição:")
print(f"  - Marcas únicas: {base_comissao['Cod_Marca'].nunique()}")
print(f"  - Cargos únicos: {base_comissao['Cod_Cargo'].nunique()}")
```

```python
# Ranking de cargos - Ordenar cargos pela comissão média (do maior para menor)
ranking_cargos = base_comissao.groupby('Descri_Cargo')['%_Comiss'].agg(['mean', 'min', 'max', 'count']).reset_index()
ranking_cargos = ranking_cargos.sort_values('mean', ascending=False).reset_index(drop=True)
ranking_cargos.columns = ['Cargo', 'Comissão Média', 'Mínima', 'Máxima', 'Qtd Marcas']
print("RANKING DE CARGOS POR COMISSÃO MÉDIA:")
print(ranking_cargos.to_string(index=False))
```

```python
# Criar grupos: "Cargos premium" (comissão > 2%), "Cargos padrão" (1-2%), "Cargos econômicos" (< 1%) e analisar distribuição por marca
def classificar_comissao(valor):
    if valor > 2:
        return 'Premium (>2%)'
    elif 1 <= valor <= 2:
        return 'Padrão (1-2%)'
    else:
        return 'Econômico (<1%)'

base_comissao['Grupo'] = base_comissao['%_Comiss'].apply(classificar_comissao)

print("DISTRIBUIÇÃO POR GRUPO DE COMISSÃO:\n")
for grupo in ['Premium (>2%)', 'Padrão (1-2%)', 'Econômico (<1%)']:
    dados_grupo = base_comissao[base_comissao['Grupo'] == grupo]
    if len(dados_grupo) > 0:
        print(f"\n{grupo}:")
        print(f"  Total de registros: {len(dados_grupo)}")
        distribuicao = dados_grupo['Descr_Marca'].value_counts()
        print(f"  Distribuição por marca:")
        for marca, qtd in distribuicao.items():
            print(f"    - {marca}: {qtd}")
            
# Resumo
print("\n\nRESUMO POR GRUPO:")
resumo = base_comissao['Grupo'].value_counts().sort_index()
for grupo, qtd in resumo.items():
    pct = (qtd / len(base_comissao)) * 100
    print(f"{grupo}: {qtd} ({pct:.1f}%)")
```

# BASE RH

```python
# Importar baseRH de setembro
base_rh = pd.read_excel('../baseRH/BASE RH_SET25.xlsx')
print("Base RH - Setembro 2025 carregada com sucesso!")
print(f"Dimensões: {base_rh.shape}")
print(f"\nColunas: {base_rh.columns.tolist()}")
print(f"\nPrimeiras linhas:")
print(base_rh.head())
```

```python
# Filtrar vendedores por marca
marca_filtro = 'PRETO'  # Altere conforme necessário
vendedores_marca = base_rh[base_rh['Descri_Marca'] == marca_filtro]
print(f"Vendedores da marca {marca_filtro}:")
print(vendedores_marca[['Matricula ', 'Descri_Marca', 'Descri_Cargo ', 'Data_Admiss ']].head(10))
print(f"\nTotal de vendedores: {len(vendedores_marca)}")
print(f"Vendedores únicos: {vendedores_marca['Matricula '].nunique()}")
```

```python
# Filtrar vendedores por cargo
cargo_filtro = 'VENDEDOR LOJA'  # Altere conforme necessário
vendedores_cargo = base_rh[base_rh['Descri_Cargo '] == cargo_filtro]
print(f"Vendedores do cargo {cargo_filtro}:")
print(vendedores_cargo[['Matricula ', 'Descri_Cargo ', 'Descri_Marca', 'Data_Admiss ']].head(10))
print(f"\nTotal de registros: {len(vendedores_cargo)}")
print(f"Vendedores únicos: {vendedores_cargo['Matricula '].nunique()}")
```

```python
# Análise de datas de admissão e demissão na Base RH
print("ANÁLISE DE ADMISSÕES E DEMISSÕES:\n")

# Converter colunas para datetime
base_rh['Data_Admiss '] = pd.to_datetime(base_rh['Data_Admiss '], errors='coerce')
base_rh['Data_Demiss'] = pd.to_datetime(base_rh['Data_Demiss'], errors='coerce')

# Maior admissão
admissao_mais_recente = base_rh['Data_Admiss '].max()
demissao_mais_recente = base_rh['Data_Demiss'].max()

print(f"Admissão mais recente: {admissao_mais_recente}")
print(f"Demissão mais recente: {demissao_mais_recente}")

# Vendedores que foram demitidos
demitidos = base_rh[base_rh['Data_Demiss'].notna()]
print(f"\nVendedores demitidos: {demitidos['Matricula '].nunique()}")
print(f"Total de registros com demissão: {len(demitidos)}")
```

```python
# Análise de vendedores por loja
vendedores_por_loja = base_rh.groupby('Descr_Loja')['Matricula '].agg(['count', 'nunique']).reset_index()
vendedores_por_loja.columns = ['Loja', 'Total Registros', 'Vendedores Únicos']
vendedores_por_loja = vendedores_por_loja.sort_values('Vendedores Únicos', ascending=False)

print("VENDEDORES POR LOJA:")
print(vendedores_por_loja.to_string(index=False))
```

```python
# Vendedor mais recente - Quem foi admitido mais recentemente?
vendedor_mais_recente = base_rh.loc[base_rh['Data_Admiss '].idxmax()]
print("VENDEDOR ADMITIDO MAIS RECENTEMENTE:")
print(f"Matrícula: {vendedor_mais_recente['Matricula ']}")
print(f"Data Admissão: {vendedor_mais_recente['Data_Admiss ']}")
print(f"Cargo: {vendedor_mais_recente['Descri_Cargo ']}")
print(f"Marca: {vendedor_mais_recente['Descri_Marca']}")
print(f"Loja: {vendedor_mais_recente['Descr_Loja']}")
```

```python
# Contar vendedores ativos - Quantos vendedores diferentes estão em base_rh?
vendedores_ativos = base_rh['Matricula '].nunique()
vendedores_inativos = base_rh[base_rh['Data_Demiss'].notna()]['Matricula '].nunique()
vendedores_sem_demissao = base_rh[base_rh['Data_Demiss'].isna()]['Matricula '].nunique()

print("CONTAGEM DE VENDEDORES:")
print(f"Total de vendedores únicos: {vendedores_ativos}")
print(f"Vendedores com demissão registrada: {vendedores_inativos}")
print(f"Vendedores sem demissão (potencialmente ativos): {vendedores_sem_demissao}")
```

```python
# Filtrar por matrícula - Todas as informações do vendedor com matrícula específica
matricula_filtro = 'MATRIC-133'  # Altere conforme necessário
vendedor = base_rh[base_rh['Matricula '] == matricula_filtro]
if len(vendedor) > 0:
    print(f"Informações do vendedor {matricula_filtro}:")
    print(vendedor[['Matricula ', 'Descri_Cargo ', 'Descri_Marca', 'Descr_Loja', 'Data_Admiss ', 'Data_Demiss']].to_string(index=False))
else:
    print(f"Vendedor {matricula_filtro} não encontrado")
```

```python
# Performance por vendedor - Análise de vendedores ativos
performance_vendedor = base_rh.groupby('Matricula ').agg({
    'Descri_Cargo ': 'first',
    'Descri_Marca': 'first',
    'Descr_Loja': 'first',
    'Data_Admiss ': 'first',
    'Data_Demiss': lambda x: x.notna().any()  # True se tem demissão
}).reset_index()

performance_vendedor.columns = ['Matricula ', 'Cargo', 'Marca', 'Loja', 'Data_Admiss ', 'Demitido']
performance_vendedor = performance_vendedor.sort_values('Matricula ')

print("PERFORMANCE DOS VENDEDORES:")
print(performance_vendedor.head(10).to_string(index=False))
print(f"\nTotal de vendedores: {len(performance_vendedor)}")
```

```python
# Análise de datas dos registros de RH
base_rh['Data_Ref'] = pd.to_datetime(base_rh['Data_Ref'], errors='coerce')

print("DATAS DISPONÍVEIS NA BASE RH:")
datas_unicas = base_rh['Data_Ref'].unique()
print(f"Data mínima: {base_rh['Data_Ref'].min()}")
print(f"Data máxima: {base_rh['Data_Ref'].max()}")

# Contar registros por data
registros_por_data = base_rh['Data_Ref'].value_counts().sort_index()
print(f"\nRegistros por data:")
print(registros_por_data)
```

```python
# Filtrar por período específico
data_inicio = '2025-09-01'
data_fim = '2025-09-30'

# Converter para datetime se ainda não foi
if base_rh['Data_Ref'].dtype == 'object':
    base_rh['Data_Ref'] = pd.to_datetime(base_rh['Data_Ref'])

periodo = base_rh[(base_rh['Data_Ref'] >= data_inicio) & (base_rh['Data_Ref'] <= data_fim)]
print(f"Registros no período de {data_inicio} a {data_fim}: {len(periodo)}")
print(f"Vendedores únicos: {periodo['Matricula '].nunique()}")
print(f"\nAmostra dos dados:")
print(periodo[['Matricula ', 'Descri_Cargo ', 'Descri_Marca', 'Data_Ref']].head(10).to_string(index=False))
```

```python
# Análise detalhada por loja
analise_loja = base_rh.groupby('Descr_Loja').agg({
    'Matricula ': 'nunique',
    'Descri_Cargo ': 'nunique',
    'Descri_Marca': 'nunique'
}).reset_index()

analise_loja.columns = ['Loja', 'Vendedores', 'Cargos', 'Marcas']
analise_loja = analise_loja.sort_values('Vendedores', ascending=False)

print("ANÁLISE DETALHADA POR LOJA:")
print(analise_loja.to_string(index=False))
```

```python
# Tempo desde admissão - Quanto tempo cada vendedor está ativo?
data_referencia = base_rh['Data_Ref'].max()
base_rh['Dias_Admitido'] = (data_referencia - base_rh['Data_Admiss ']).dt.days
base_rh['Anos_Admitido'] = base_rh['Dias_Admitido'] / 365.25

# Análise por vendedor único
tempo_admissao = base_rh.drop_duplicates('Matricula ')[['Matricula ', 'Data_Admiss ', 'Dias_Admitido', 'Anos_Admitido']].sort_values('Dias_Admitido', ascending=False)

print("TEMPO DESDE ADMISSÃO:")
print(f"Data de referência: {data_referencia}")
print(f"\nVendedores com maior tempo na casa:")
print(tempo_admissao.head(10).to_string(index=False))

print(f"\nVendedores mais recentes:")
print(tempo_admissao.tail(5).to_string(index=False))
```

```python
# Marcas por loja - Quais marcas cada loja vende?
marcas_por_loja = base_rh.groupby('Descr_Loja')['Descri_Marca'].unique().reset_index()
marcas_por_loja.columns = ['Loja', 'Marcas']

print("MARCAS VENDIDAS POR LOJA:\n")
for idx, row in marcas_por_loja.iterrows():
    marcas = ', '.join(sorted(row['Marcas']))
    print(f"{row['Loja']}: {marcas}")
```

```python
# Análise de sazonalidade - Há padrão nas datas de admissão?
base_rh['Mes_Admiss'] = base_rh['Data_Admiss '].dt.month
base_rh['Ano_Admiss'] = base_rh['Data_Admiss '].dt.year

admissoes_por_mes = base_rh.drop_duplicates('Matricula ').groupby('Mes_Admiss').size().reset_index(name='Quantidade')
admissoes_por_mes['Mês'] = admissoes_por_mes['Mes_Admiss'].map({
    1: 'Janeiro', 2: 'Fevereiro', 3: 'Março', 4: 'Abril', 5: 'Maio', 6: 'Junho',
    7: 'Julho', 8: 'Agosto', 9: 'Setembro', 10: 'Outubro', 11: 'Novembro', 12: 'Dezembro'
})

print("ADMISSÕES POR MÊS (Sazonalidade):")
print(admissoes_por_mes[['Mês', 'Quantidade']].to_string(index=False))
```

```python
# Produtividade por tempo de casa - Vendedores mais recentes vs antigos
# Para isso, precisamos de dados de vendas. Por enquanto, segmentamos por tempo de casa
base_rh['Segmento_Tempo'] = pd.cut(base_rh['Dias_Admitido'], 
                                    bins=[0, 365, 730, 1095, 99999],
                                    labels=['< 1 ano', '1-2 anos', '2-3 anos', '> 3 anos'])

analise_tempo = base_rh.drop_duplicates('Matricula ').groupby('Segmento_Tempo').agg({
    'Matricula ': 'count',
    'Descri_Cargo ': 'unique'
}).reset_index()

print("DISTRIBUIÇÃO DE VENDEDORES POR TEMPO DE CASA:")
print(analise_tempo[['Segmento_Tempo', 'Matricula ']].to_string(index=False))
print("\nNota: Para análise de produtividade, será necessário cruzar com dados de vendas.")
```

```python
# Análise de turnover - Quantos vendedores deixaram a empresa?
demissoes = base_rh[base_rh['Data_Demiss'].notna()]
vendedores_demitidos = demissoes['Matricula '].nunique()
total_vendedores = base_rh['Matricula '].nunique()
taxa_turnover = (vendedores_demitidos / total_vendedores) * 100

print("ANÁLISE DE TURNOVER:")
print(f"Total de vendedores únicos: {total_vendedores}")
print(f"Vendedores demitidos: {vendedores_demitidos}")
print(f"Taxa de turnover: {taxa_turnover:.2f}%")

print("\nVendedores demitidos:")
demitidos_detalhes = demissoes.drop_duplicates('Matricula ')[['Matricula ', 'Descri_Cargo ', 'Data_Admiss ', 'Data_Demiss']].sort_values('Data_Demiss')
print(demitidos_detalhes.to_string(index=False))
```

```python
# Correlação marca-loja - Algumas marcas vendem melhor em certas lojas?
correlacao_marca_loja = base_rh.groupby(['Descr_Loja', 'Descri_Marca'])['Matricula '].nunique().reset_index()
correlacao_marca_loja.columns = ['Loja', 'Marca', 'Vendedores']
correlacao_marca_loja = correlacao_marca_loja.sort_values(['Loja', 'Vendedores'], ascending=[True, False])

print("CORRELAÇÃO MARCA-LOJA (Preferência Geográfica):")
print("\nVendedores por marca em cada loja:")
for loja in correlacao_marca_loja['Loja'].unique():
    print(f"\n{loja}:")
    dados_loja = correlacao_marca_loja[correlacao_marca_loja['Loja'] == loja]
    for _, row in dados_loja.iterrows():
        print(f"  {row['Marca']}: {row['Vendedores']} vendedores")
```

```python
# Filtrar por marca - Todas as vendas das marcas especificadas
marcas_filtro = ['PRETO', 'VERMELHO']  # Altere conforme necessário
vendas_marca = base_venda[base_venda['Descr_Marca'].isin(marcas_filtro)]

print(f"Vendas das marcas {', '.join(marcas_filtro)}:")
print(f"Total de vendas: {len(vendas_marca)}")
print(f"Total de unidades: {vendas_marca['Vlr _Venda'].sum():.2f}")

print("\nVendas por marca:")
vendas_por_marca = vendas_marca.groupby('Descr_Marca')['Vlr _Venda'].agg(['sum', 'count', 'mean']).reset_index()
vendas_por_marca.columns = ['Marca', 'Total', 'Quantidade', 'Média']
print(vendas_por_marca.to_string(index=False))
```

```python
# Filtrar por vendedor - Todas as vendas de um vendedor específico
vendedor_filtro = 'MATRIC-56'  # Altere conforme necessário
vendas_vendedor = base_venda[base_venda['Matricula'] == vendedor_filtro]

print(f"Vendas do vendedor {vendedor_filtro}:")
print(f"Total de vendas: {len(vendas_vendedor)}")
print(f"Total de unidades: {vendas_vendedor['Vlr _Venda'].sum():.2f}")
print(f"Média por venda: {vendas_vendedor['Vlr _Venda'].mean():.2f}")

print(f"\nVendas por data:")
vendas_por_data = vendas_vendedor.groupby('Date_Ref')['Vlr _Venda'].agg(['sum', 'count']).reset_index()
vendas_por_data.columns = ['Data', 'Total', 'Quantidade']
print(vendas_por_data.to_string(index=False))
```

```python
# Contar vendedores únicos - Quantos vendedores diferentes aparecem?
vendedores_unicos = base_venda['Matricula'].nunique()
print(f"Total de vendedores únicos na base de vendas: {vendedores_unicos}")

# Análise detalhada
print(f"\nTop 10 vendedores por número de vendas:")
top_vendedores = base_venda['Matricula'].value_counts().head(10).reset_index()
top_vendedores.columns = ['Matricula', 'Quantidade_Vendas']
print(top_vendedores.to_string(index=False))
```

```python
# Contar lojas únicas - Quantas lojas diferentes venderam?
lojas_unicas = base_venda['Descr_Loja'].nunique()
print(f"Total de lojas únicas na base de vendas: {lojas_unicas}")

# Análise detalhada
print(f"\nLojas e suas quantidades de vendas:")
vendas_por_loja = base_venda['Descr_Loja'].value_counts().reset_index()
vendas_por_loja.columns = ['Loja', 'Quantidade_Vendas']
print(vendas_por_loja.to_string(index=False))
```

```python
# Contar marcas únicas - Quantas marcas foram vendidas?
marcas_unicas = base_venda['Descr_Marca'].nunique()
print(f"Total de marcas únicas na base de vendas: {marcas_unicas}")

# Lista de marcas
print(f"\nMarcas vendidas:")
marcas_lista = base_venda['Descr_Marca'].unique()
for marca in sorted(marcas_lista):
    print(f"  - {marca}")
```

```python
# Total de unidades vendidas - Somar todos os Vlr_Venda (quantidade total)
total_vendido = base_venda['Vlr _Venda'].sum()
total_registros = len(base_venda)
media_por_venda = base_venda['Vlr _Venda'].mean()

print("RESUMO GERAL DE VENDAS:")
print(f"Total de registros de venda: {total_registros:,}")
print(f"Total de unidades vendidas: {total_vendido:,.2f}")
print(f"Média por venda: {media_por_venda:.2f}")
print(f"Máximo por venda: {base_venda['Vlr _Venda'].max():.2f}")
print(f"Mínimo por venda: {base_venda['Vlr _Venda'].min():.2f}")
```

```python
# Total vendido por marca - PRETO vs VERMELHO: quem vendeu mais unidades?
# Total vendido por loja - Ranking de lojas por volume de vendas
# Total vendido por vendedor - Qual vendedor movimentou mais unidades?

print("=" * 60)
print("TOTAL VENDIDO POR MARCA")
print("=" * 60)
total_marca = base_venda.groupby('Descr_Marca')['Vlr _Venda'].agg(['sum', 'count', 'mean']).reset_index()
total_marca.columns = ['Marca', 'Total', 'Quantidade', 'Média']
total_marca = total_marca.sort_values('Total', ascending=False)
print(total_marca.to_string(index=False))

print("\n" + "=" * 60)
print("RANKING DE LOJAS POR VOLUME DE VENDAS")
print("=" * 60)
total_loja = base_venda.groupby('Descr_Loja')['Vlr _Venda'].agg(['sum', 'count', 'mean']).reset_index()
total_loja.columns = ['Loja', 'Total', 'Quantidade', 'Média']
total_loja = total_loja.sort_values('Total', ascending=False)
print(total_loja.to_string(index=False))

print("\n" + "=" * 60)
print("RANKING DE VENDEDORES POR VOLUME DE VENDAS")
print("=" * 60)
total_vendedor = base_venda.groupby('Matricula')['Vlr _Venda'].agg(['sum', 'count', 'mean']).reset_index()
total_vendedor.columns = ['Matricula', 'Total', 'Quantidade', 'Média']
total_vendedor = total_vendedor.sort_values('Total', ascending=False)
print(total_vendedor.head(15).to_string(index=False))
```

```python
# Análise de concentração - Qual percentual das vendas vem dos top 20% dos vendedores?
# Calcular total por vendedor
vendedor_total = base_venda.groupby('Matricula')['Vlr _Venda'].sum().sort_values(ascending=False)

# Calcular total geral
total_geral = vendedor_total.sum()

# Calcular quantos vendedores representam 20%
total_vendedores = len(vendedor_total)
top_20_pct = int(total_vendedores * 0.20)
if top_20_pct == 0:
    top_20_pct = 1

# Pegar os top vendedores (20%)
top_vendedores_20 = vendedor_total.head(top_20_pct)
total_top_20 = top_vendedores_20.sum()
percentual_vendas = (total_top_20 / total_geral) * 100

print("ANÁLISE DE CONCENTRAÇÃO DE VENDAS:")
print(f"Total de vendedores: {total_vendedores}")
print(f"Top 20% de vendedores: {top_20_pct}")
print(f"Total vendido pelos top 20%: {total_top_20:.2f}")
print(f"Percentual do total: {percentual_vendas:.2f}%")

print(f"\nTop {top_20_pct} vendedores (20%):")
top_20_detalhes = top_vendedores_20.reset_index()
top_20_detalhes.columns = ['Matricula', 'Total']
top_20_detalhes['Percentual'] = (top_20_detalhes['Total'] / total_geral * 100).round(2)
print(top_20_detalhes.to_string(index=False))
```
